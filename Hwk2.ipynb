**(c)** (25 points) Perform a 5-folds cross validation with the goal of measuring the performance, in terms of F1-score, of two competing models:

- Using `age`, `currentSmoker`, `totChol`, `sysBP`, `diaBP`, `BMI`, `heartRate`, and `glucose` as the predictor variables, and `TenYearCHD` as the target variable build a logistic regression model under the 5-folds cross validation framework. Compute and store the F1-score for each iteration.
- Using `age`, `currentSmoker`, `totChol`, `BMI`, `heartRate`, and `glucose` as the predictor variables, and `TenYearCHD` as the target variable build a logistic regression model under the 5-folds cross validation framework. Compute and store the F1-score for each iteration.

Use 25% as threshold to change the likelihoods to labels. Make sure to scale the input variables of both models to 0-1 range (see [`MinMaxScaler`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.MinMaxScaler.html)) before you run the 5-fold cross validation framework. Also, you can use the [`f1_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html) function to compute the F1-score.

```python
# Your code here
from sklearn.model_selection import KFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

X = heart[['age', 'currentSmoker', 'totChol', 'sysBP', 'diaBP', 'BMI', 'heartRate', 'glucose']]
Y = heart['TenYearCHD']

X_1 = heart[['age', 'currentSmoker', 'totChol', 'BMI', 'heartRate', 'glucose']]
Y_1 = heart['TenYearCHD']

kfold = KFold(n_splits=5, shuffle=True, random_state=42)

split = kfold.split(X, Y)
split_1 = kfold.split(X_1, Y_1)

f1_scores = []
f1_scores_1 = []

for train_index, test_index in split:
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    Y_train, Y_test = Y.iloc[train_index], Y.iloc[test_index]
    
    pipeline = make_pipeline(MinMaxScaler(), LogisticRegression())
    pipeline.fit(X_train, Y_train)
    
    probabilities = pipeline.predict_proba(X_test)[:, 1]
    predictions = (probabilities >= 0.25).astype(int)
    
    f1_scores.append(f1_score(Y_test, predictions))
    
for train_index, test_index in split_1:
    X_train, X_test = X_1.iloc[train_index], X_1.iloc[test_index]
    Y_train, Y_test = Y_1.iloc[train_index], Y_1.iloc[test_index]
    
    pipeline = make_pipeline(MinMaxScaler(), LogisticRegression())
    pipeline.fit(X_train, Y_train)
    
    probabilities = pipeline.predict_proba(X_test)[:, 1]
    predictions = (probabilities >= 0.25).astype(int)
    
    f1_scores_1.append(f1_score(Y_test, predictions))

print("F1 scores with all features:", f1_scores)
print("F1 scores with selected features:", f1_scores_1)

print("Average F1 score with all features:", sum(f1_scores) / len(f1_scores))
print("Average F1 score with selected features:", sum(f1_scores_1) / len(f1_scores_1))
```
