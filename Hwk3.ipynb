```python
# (a) Read the csv data file and create the car_price data-frame

import pandas as pd

car_price = pd.read_csv("CarPrice_Assignment.csv")
car_price.head()
```

```python
# (b) Split the data into train (80%) and test (20%)
from sklearn.model_selection import train_test_split

train, test = train_test_split(car_price, test_size=0.2, random_state=42)
```

```python
# (c) LASSO variable selection repeated 1000 times

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LassoCV, Lasso
import numpy as np

x_train = train[['wheelbase','enginesize','compressionratio','horsepower','peakrpm','citympg','highwaympg']]
y_train = train['price']

coef_df = pd.DataFrame(columns=['features', 'coefficient'])


for i in range(1000):
    lasso = make_pipeline(StandardScaler(),
                      LassoCV(cv=5, n_jobs=-1))
    lasso.fit(x_train, y_train)


    lasso_md = make_pipeline(StandardScaler(),
                            Lasso(alpha=lasso.named_steps['lassocv'].alpha_))
    lasso_md.fit(x_train, y_train)
        
    coef_df = pd.concat([coef_df, pd.DataFrame({
        'features': x_train.columns,
        'coefficient': lasso_md.named_steps['lasso'].coef_
    })], ignore_index=True)
```

```python
zeros = coef_df[coef_df['coefficient'] == 0].groupby('features').size()
zeros
```

```python
# (d) Linear regression model and RMSE on test dataset

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_error

x_train = train[['wheelbase','enginesize','compressionratio','horsepower','peakrpm','citympg']]
y_train = train['price']

x_test = test[['wheelbase','enginesize','compressionratio','horsepower','peakrpm','citympg']]
y_test = test['price']

lr = make_pipeline(StandardScaler(), 
                   LinearRegression())

lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("RMSE on test dataset:", rmse)
```

```python
# (e) Ridge regression model (optimal lambda selection repeated 100 times) and RMSE on test dataset
from sklearn.linear_model import Ridge, RidgeCV
ridge_alpha = [0.001 ,0.01, 0.1, 1, 10, 100]

lambda_optimal = []

for i in range(100):
    ridge = make_pipeline(StandardScaler(), 
                         RidgeCV(alphas=ridge_alpha, cv=5))
    
    ridge.fit(x_train, y_train)
    lambda_optimal.append(ridge.named_steps['ridgecv'].alpha_)

ridge_lambda_df = pd.DataFrame(lambda_optimal, columns=['lambda_optimal'])
ridge_lambda_df.head()
```

```python
print(ridge_lambda_df['lambda_optimal'].mode())
```

```python
last_ridge = make_pipeline(StandardScaler(), 
                           Ridge(alpha=.001))

last_ridge.fit(x_train, y_train)

y_pred_ridge = last_ridge.predict(x_test)
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
print("RMSE on test dataset:", rmse_ridge)
```

Using the same `train` and `test` datasets created in Exercise 5(b), with `wheelbase`, `enginesize`, `compressionratio`, `horsepower`, `peakrpm`, `citympg`, and `highwaympg` as the predictor variables and `price` as the target variable, answer the following. (a) (10 points) Build an Elastic Net regression model. Use `ElasticNetCV` with 5-folds to estimate the optimal `alpha` (regularization strength) and `l1_ratio`, considering `l1_ratio` values of [0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 1]. Use `GridSearchCV` with 5-folds to estimate the optimal `epsilon` from the set [1.1, 1.35, 1.5, 1.75, 2] and `alpha` from the set [0.0001, 0.001, 0.01, 0.1, 1].
```python
# (a) Elastic Net model (optimal alpha and l1_ratio via ElasticNetCV) and RMSE on test dataset
from sklearn.linear_model import ElasticNetCV, ElasticNet

x_train_1 = train[['wheelbase','enginesize','compressionratio','horsepower','peakrpm','citympg','highwaympg']]
y_train_1 = train['price']

x_test_1 = test[['wheelbase','enginesize','compressionratio','horsepower','peakrpm','citympg','highwaympg']]
y_test_1 = test['price']


elastic_net_cv = make_pipeline(StandardScaler(), 
                               ElasticNetCV(l1_ratio=[.1, .3, .5, .7, .9, .95, 1], cv=5, random_state=42))

elastic_net_cv.fit(x_train_1, y_train_1)

alpha_optimal = elastic_net_cv.named_steps['elasticnetcv'].alpha_
l1_ratio_optimal = elastic_net_cv.named_steps['elasticnetcv'].l1_ratio_

print("Optimal alpha:", alpha_optimal)
print("Optimal l1_ratio:", l1_ratio_optimal)

```

```python
elastic_net = make_pipeline(StandardScaler(), 
                           ElasticNet(alpha=alpha_optimal, l1_ratio=l1_ratio_optimal))
elastic_net.fit(x_train_1, y_train_1)

y_pred_1 = elastic_net.predict(x_test_1)

from sklearn.metrics import mean_squared_error
rmse_1 = np.sqrt(mean_squared_error(y_test_1, y_pred_1))
print("RMSE on test dataset:", rmse_1)
```

```python
# (b) Huber Regression model (optimal epsilon and alpha via GridSearchCV) and RMSE on test dataset
from sklearn.linear_model import HuberRegressor
from sklearn.model_selection import GridSearchCV

huber = HuberRegressor()

alpha_values = [0.0001, 0.001, 0.01, 0.1, 1]
epsilon_values = [1.1, 1.35, 1.5, 1.75,2]

param_grid = {
    'alpha': alpha_values,
    'epsilon': epsilon_values
}

gridsearch = GridSearchCV(huber, param_grid, cv=5, n_jobs=-1)
gridsearch.fit(x_train_1, y_train_1)
best_huber = gridsearch.best_estimator_

y_pred = best_huber.predict(x_test_1)
rmse = np.sqrt(mean_squared_error(y_test_1, y_pred))
print("RMSE on test dataset:", rmse)
```
