### (a) (5 points)

Using the `pandas` library, read the csv data file and create a data-frame called `heart`. Remove the observations with missing values.

```python
# Your code here

import pandas as pd 

heart = pd.read_csv("framingham.csv")
heart.dropna(inplace=True)
heart.reset_index(drop=True, inplace=True)
heart.head()

heart.isnull().sum()
```

### (b) (60 points)

Based on the different interactions that we have had so far with this data file, it seems that `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the most important predictor variables when it comes to predicting the likelihood of `TenYearCHD`.

Using `TenYearCHD` as the target variable and `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` as the input variables, perform a **5-fold stratified cross-validation** (i.e., use `StratifiedKFold` with `n_splits = 5`) to evaluate each of the following models. For every fold, train the model on the four training folds and use it to predict the risk of coronary heart disease for the patients in the held-out fold. Using 10% as the cutoff value to convert the predicted probabilities into class labels, compute the accuracy and recall for that fold. After all 5 folds are complete, report the **average accuracy** and **average recall** across the 5 folds for each of the following models:

(i) `RandomForestClassifier` (with `n_estimators = 500` and `max_depth = 3`) in which `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the predictor variables, and `TenYearCHD` is the target variable.

(ii) `ExtraTreesClassifier` (with `n_estimators = 500` and `max_depth = 3`) in which `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the predictor variables, and `TenYearCHD` is the target variable.

(iii) `AdaBoostClassifier` (with `n_estimators = 500`, `max_depth = 3`, and `learning_rate = 0.01`) in which `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the predictor variables, and `TenYearCHD` is the target variable.

(iv) `GradientBoostingClassifier` (with `n_estimators = 500`, `max_depth = 3`, and `learning_rate = 0.01`) in which `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the predictor variables, and `TenYearCHD` is the target variable.

(v) `HistGradientBoostingClassifier` (with `max_depth = 3` and `learning_rate = 0.01`) in which `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the predictor variables, and `TenYearCHD` is the target variable.

(vi) `LGBMClassifier` (with `n_estimators = 500`, `max_depth = 3`, and `learning_rate = 0.01`) in which `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the predictor variables, and `TenYearCHD` is the target variable.

Based on the average accuracy and average recall across the 5 folds, what model would you use to predict `TenYearCHD`?

```python
# Your code here
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
from sklearn.metrics import recall_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from lightgbm import LGBMClassifier


x = heart[[ "age", "totChol", "sysBP", "BMI", "heartRate", "glucose" ]]
y = heart["TenYearCHD"]

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
rf_recall = []
rf_accuracy = []

for i, (train_index, test_index) in enumerate(skf.split(x, y)):
    x_train, x_test = x.iloc[train_index], x.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    rf = RandomForestClassifier(n_estimators=500, 
                                max_depth=3, 
                                random_state=42,
                                n_jobs=-1)
    rf.fit(x_train, y_train)
    y_pred = rf.predict_proba(x_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    rf_recall.append(recall_score(y_test, y_pred))
    rf_accuracy.append(accuracy_score(y_test, y_pred))

print("Random Forest average Recall: ", sum(rf_recall)/len(rf_recall))
print("Random Forest average Accuracy: ", sum(rf_accuracy)/len(rf_accuracy))
```

```python
et_recall = []
et_accuracy = []

for i, (train_index, test_index) in enumerate(skf.split(x, y)):
    x_train, x_test = x.iloc[train_index], x.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    et = ExtraTreesClassifier(n_estimators=500, 
                              max_depth=3, 
                              random_state=42,
                              n_jobs=-1)
    et.fit(x_train, y_train)
    y_pred = et.predict_proba(x_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    et_recall.append(recall_score(y_test, y_pred))
    et_accuracy.append(accuracy_score(y_test, y_pred))

print("Extra Trees average Recall: ", sum(et_recall)/len(et_recall))
print("Extra Trees average Accuracy: ", sum(et_accuracy)/len(et_accuracy))
```

```python
ada_recall = []
ada_accuracy = []

for i, (train_index, test_index) in enumerate(skf.split(x, y)):
    x_train, x_test = x.iloc[train_index], x.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    ada = AdaBoostClassifier(n_estimators=500, 
                             estimator=DecisionTreeClassifier(max_depth=3),
                             learning_rate=0.01,
                             random_state=42)
    ada.fit(x_train, y_train)
    y_pred = ada.predict_proba(x_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    ada_recall.append(recall_score(y_test, y_pred))
    ada_accuracy.append(accuracy_score(y_test, y_pred))

print("AdaBoost average Recall: ", sum(ada_recall)/len(ada_recall))
print("AdaBoost average Accuracy: ", sum(ada_accuracy)/len(ada_accuracy))
```

```python
gb_recall = []
gb_accuracy = []

for i, (train_index, test_index) in enumerate(skf.split(x, y)):
    x_train, x_test = x.iloc[train_index], x.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    gb = GradientBoostingClassifier(n_estimators=500, 
                                    max_depth=3, 
                                    learning_rate=0.01,
                                    random_state=42)
    gb.fit(x_train, y_train)
    y_pred = gb.predict_proba(x_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    gb_recall.append(recall_score(y_test, y_pred))
    gb_accuracy.append(accuracy_score(y_test, y_pred))

print("Gradient Boosting average Recall: ", sum(gb_recall)/len(gb_recall))
print("Gradient Boosting average Accuracy: ", sum(gb_accuracy)/len(gb_accuracy))
```

```python
hist_gb_recall = []
hist_gb_accuracy = []

for i, (train_index, test_index) in enumerate(skf.split(x, y)):
    x_train, x_test = x.iloc[train_index], x.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    hist_gb = HistGradientBoostingClassifier(max_depth=3, 
                                             learning_rate=0.01,
                                             random_state=42)
    hist_gb.fit(x_train, y_train)
    y_pred = hist_gb.predict_proba(x_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    hist_gb_recall.append(recall_score(y_test, y_pred))
    hist_gb_accuracy.append(accuracy_score(y_test, y_pred))
    
print("Hist Gradient Boosting average Recall: ", sum(hist_gb_recall)/len(hist_gb_recall))
print("Hist Gradient Boosting average Accuracy: ", sum(hist_gb_accuracy)/len(hist_gb_accuracy))
```

```python
lgbm_recall = []
lgbm_accuracy = []

for i, (train_index, test_index) in enumerate(skf.split(x, y)):
    x_train, x_test = x.iloc[train_index], x.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    lgbm = LGBMClassifier(n_estimators=500, 
                          max_depth=3, 
                          learning_rate=0.01,
                          random_state=42,
                          verbose=-1,
                          n_jobs=-1)
    lgbm.fit(x_train, y_train)
    y_pred = lgbm.predict_proba(x_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    lgbm_recall.append(recall_score(y_test, y_pred))
    lgbm_accuracy.append(accuracy_score(y_test, y_pred))
    
print("LightGBM average Recall: ", sum(lgbm_recall)/len(lgbm_recall))
print("LightGBM average Accuracy: ", sum(lgbm_accuracy)/len(lgbm_accuracy))
```

```python
print("Random Forest average Recall: ", sum(rf_recall)/len(rf_recall))
print("Random Forest average Accuracy: ", sum(rf_accuracy)/len(rf_accuracy))
print("Extra Trees average Recall: ", sum(et_recall)/len(et_recall))
print("Extra Trees average Accuracy: ", sum(et_accuracy)/len(et_accuracy))
print("AdaBoost average Recall: ", sum(ada_recall)/len(ada_recall))
print("AdaBoost average Accuracy: ", sum(ada_accuracy)/len(ada_accuracy))
print("Gradient Boosting average Recall: ", sum(gb_recall)/len(gb_recall))
print("Gradient Boosting average Accuracy: ", sum(gb_accuracy)/len(gb_accuracy))
print("Hist Gradient Boosting average Recall: ", sum(hist_gb_recall)/len(hist_gb_recall))
print("Hist Gradient Boosting average Accuracy: ", sum(hist_gb_accuracy)/len(hist_gb_accuracy))
print("LightGBM average Recall: ", sum(lgbm_recall)/len(lgbm_recall))
print("LightGBM average Accuracy: ", sum(lgbm_accuracy)/len(lgbm_accuracy))
```
